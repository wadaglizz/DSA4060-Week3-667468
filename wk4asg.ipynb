```python
watched_movies = ratings["movie_id"]

recommendations = ranked_movies[
    ~ranked_movies["movie_id"].isin(watched_movies)
]

recommendations[
    ["title", "recommendation_score"]
]
```

RECOMMENDATION BREAKDOWN

Space Warriors is recommended because it contains both Action and
SciFi, which are User A's two strongest preferences.

City Detectives is recommended because it contains Action, which is
the user's strongest individual preference.

Hidden Truth receives a smaller positive score because it contains
Drama, which is a relatively weak preference for User A.


```python
def recommend_movies(movies, ratings, features, n=3):
    user_movies = ratings.merge(
        movies,
        on="movie_id"
    )

    rated_features = user_movies[features]

    weighted_features = rated_features.mul(
        user_movies["rating"],
        axis=0
    )

    user_profile = weighted_features.sum()

    user_profile = user_profile / user_profile.sum()

    scores = (
        movies[features]
        .mul(user_profile)
        .sum(axis=1)
    )

    results = movies.copy()

    results["recommendation_score"] = scores

    watched = ratings["movie_id"]

    results = results[
        ~results["movie_id"].isin(watched)
    ]

    return results.sort_values(
        "recommendation_score",
        ascending=False
    ).head(n)
```

```python
recommend_movies(
    movies,
    ratings,
    features
)
```

```python
user_b_ratings = pd.DataFrame({
    "movie_id": [2, 4, 6, 9],
    "rating": [5, 4, 5, 4]
})

recommend_movies(
    movies,
    user_b_ratings,
    features
)
```

User A and User B receive different recommendations because their ratings
produce different user preference profiles.

User A strongly prefers Action and SciFi, while User B strongly prefers
Romance, Drama, and Comedy.

Since recommendation scores are calculated from these profiles, movies containing the preferred features receive higher scores
for each user.
